```python
#%load_ext autoreload
#%autoreload 2

import pandas as pd
from wearable_affect.data import PROJECT_ROOT
from wearable_affect.models import make_dummy, make_logreg, make_lightgbm
from wearable_affect.tracking import run_loso_experiment #, summarise

features = pd.read_parquet(PROJECT_ROOT / "data" / "processed" / "features.parquet")
settings = {"window_s": 60, "step_s": 30, "threshold": 0.5, "features": "neurokit_v1"}

for name, make_model in [("dummy", make_dummy), ("logreg", make_logreg), ("lightgbm", make_lightgbm)]:
    results = run_loso_experiment(f"baseline-{name}", features, make_model, params=settings)
    print(name, results["macro_f1"].mean().round(3))
```
