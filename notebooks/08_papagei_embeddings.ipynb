```python

import time
from wearable_affect.data import DEFAULT_DATA_DIR, PROJECT_ROOT, list_subjects, load_subject
from wearable_affect.windows import make_windows
from wearable_affect.papagei import load_papagei, build_embedding_table

windows = []
for sid in list_subjects(DEFAULT_DATA_DIR):
    windows.extend(make_windows(load_subject(DEFAULT_DATA_DIR, sid)))

model, device = load_papagei()
start = time.time()
embeddings = build_embedding_table(windows, model, device)
print(f"{embeddings.shape} on {device} in {time.time() - start:.1f} s")

embeddings.to_parquet(PROJECT_ROOT / "data" / "processed" / "papagei_embeddings.parquet", index=False)
```
