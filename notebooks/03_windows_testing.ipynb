```python
%load_ext autoreload
%autoreload 2

from collections import Counter
from wearable_affect.data import DEFAULT_DATA_DIR, list_subjects, load_subject
from wearable_affect.windows import make_windows

all_windows = []
for sid in list_subjects(DEFAULT_DATA_DIR):
    windows = make_windows(load_subject(DEFAULT_DATA_DIR, sid))
    counts = Counter(w.target for w in windows)
    print(f"{sid}: {len(windows):4d} windows  non-stress={counts[0]:3d}  stress={counts[1]:3d}")
    all_windows.extend(windows)

total = Counter(w.target for w in all_windows)
print(f"\nTotal: {len(all_windows)} windows, stress fraction = {total[1] / len(all_windows):.1%}")
```
